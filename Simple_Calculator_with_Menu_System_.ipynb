
Simple Calculator with Menu System

```python
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

while True:
    print("\n===== Simple Calculator =====")
    print("\n_____ Menu _____")
    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")
    print("5. Modulus")
    print("6. Power")
    print("7. Exit")

    choice_num = input("Enter your choice (1-7): ")

    if choice_num== "1":
        result = num1 + num2
        print("=======================")
        print("Result:", result)
        print("=======================")

    elif choice_num == "2":
        result = num1 - num2
        print("====================")
        print("Result:", result)
        print("====================")

    elif choice_num == "3":
        result = num1 * num2
        print("======================")
        print("Result:", result)
        print("======================")

    elif choice_num== "4":
        if num2 != 0:
            result = num1 / num2
            print("====================")
            print("Result:", result)
            print("====================")
        else:
            print("=====================")
            print("Error: Division by zero is not allowed.")
            print("=====================")

    elif choice_num== "5":
        if num2 != 0:
            result = num1 % num2
            print("=====================")
            print("Result:", result)
            print("======================")
        else:
            print("======================")
            print("Error: Modulus by zero is not allowed.")
            print("======================")

    elif choice_num == "6":
        result = num1 ** num2
        print("========================")
        print("Result:", result)
        print("=========================")

    if choice_num == "7":
        print("Calculator closed. Thank you!")
        break

    if choice_num not in ["1", "2", "3", "4", "5", "6"]:
        print("Invalid choice! Please choose from 1 to 7.")
        continue
```
